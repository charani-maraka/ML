```python
sd=pd.read_csv('Salary_dataset.csv')
sd = sd.rename(columns={'Unnamed: 0': 'sno'})
sd
```

```python
cov=sd['YearsExperience'].cov(sd['Salary'])
cov
```

The covariance  is 76106.30344827585 which is positive sign we can say that it is directly proportional
* If YearsExperience increases the Salary increases
* If YearsExperience decreases the Salary decreases

```python
X=sd[['YearsExperience']]
y=sd['Salary']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_train_pred=model.predict(X_train)
```

```python
y_test_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,root_mean_squared_error,r2_score
```

```python
mean_squared_error(y_train,y_train_pred)
```

```python
mean_squared_error(y_test,y_test_pred)
```

```python
mean_absolute_error(y_train,y_train_pred)
```

```python
mean_absolute_error(y_test,y_test_pred)
```

```python
root_mean_squared_error(y_train,y_train_pred)
```

```python
root_mean_squared_error(y_test,y_test_pred)
```

```python
r2_score(y_train,y_train_pred)
```

```python
r2_score(y_test,y_test_pred)
```
