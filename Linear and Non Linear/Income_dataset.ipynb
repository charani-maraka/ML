```python
import pandas as pd
# import numpy as np
# import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sd=pd.read_csv('Income_dataset.csv')
sd = sd.rename(columns={'Unnamed: 0': 'sno'})
sd
```

```python
cov=sd['YearsExperience'].cov(sd['Salary'])
cov
```

The covariance  is 76106.30344827585 which is positive sign we can say that it is directly proportional
* If YearsExperience increases the Salary increases
* If YearsExperience decreases the Salary decreases

```python
sd_clean=sd.drop(columns=['sno'])
corr=sd_clean.corr()
corr
```

The correlation of the  dataset 
The YearsExperience is directly proportional to the Salary
* If YearsExperience increases the Salary increases
* If YearsExperience decreases the Salary decreases
It is Strongly correlated because its value is near to 1
