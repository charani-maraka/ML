```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(1)
x=np.random.randint(50,200,size=(100,1))
y=np.random.randint(50,200,size=(100,1))
# y=2*x+10+np.random.normal(1,17,size=(100,1))
x.shape,y.shape
```

```python
x.ndim,y.ndim
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
from sklearn.model_selection import train_test_split      # Data Splitting    
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.model_selection import LinearRegression
model=LinearRegression()    # model creation
```

```python
# model training
model.fit(x_train,y_train)
```

```python
model.intercept_
```

```python
model.coef_
```

```python
model.score(x_train,y_train)           # Training score
```

```python
model.score(x_test,y_test)             # Test score
```
