```python
import pandas as pd
```

```python
df=pd.DataFrame({'x':[2,3,4,5],'y':[50,100,150,200]})
```

```python
x_mean=df['x'].mean()
```

```python
y_mean=df['y'].mean()
```

```python
df
```

```python
df['x-x_mean']=df['x']-x_mean
df['y-y_mean']=df['y']-y_mean
```

```python
df['(x-x_mean)(y-y_mean)']=df['x-x_mean']*df['y-y_mean']
```

```python
df['(x-x_mean)sqr']=df['x-x_mean']**2
```

```python
slope=df['(x-x_mean)(y-y_mean)'].sum()/df['x-x_mean)sqr'].sum()
slope
```

```python
intercept=y_mean-slope*x_mean
intercept 
```
