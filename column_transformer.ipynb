```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
import pandas as pd
```

```python
df=pd.read_csv('loan_data.csv')
```

```python
df.head
```

```python
df.info()
```

```python
X=df.drop(columns="loan_status")
y=df['loan_status']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(
    X,y,train_size=0.8,random_state=42
)
```

```python
xtrain['person_education'].unique
```

```python
preprocessing=ColumnTransformer(
    transformers=[
        ('ordinal_Encoder',OrdinalEncoder(
            categories=[['High School','Bachelor','Master','Associate','Doctorate']],
            handle_unknown='error'),['person_education']),
        ('oneHot_Encoder',OneHotEncoder(
            sparse_output=False,
            handle_unknown='ignore'),
         ['person_gender','person_home_ownership','loan_intent','previous_loan_defaults_on_file'])
    ],
    remainder="passthrough"
)
```

```python
preprocessing.fit_transform(xtrain)
```

```python
preprocessing.get_feature_names_out()
```

```python
preprocessing.transform(xtest)
```
